```python
import numpy as np
import time

arr = np.arange(1_000_000)

start = time.time()
loop_result = [x * 2 for x in arr]
loop_time = time.time() - start

start = time.time()
vec_result = arr * 2
vec_time = time.time() - start

print(f"Loop time: {loop_time:.4f}s")
print(f"Vectorized time: {vec_time:.4f}s")
```
